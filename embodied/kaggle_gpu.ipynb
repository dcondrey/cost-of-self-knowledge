# Self-measurement irreducibility on a GPU (Kaggle)

**First: Settings (right panel) -> Accelerator -> GPU.** Then Run All.

Runs the hysteresis/floor test on the GPU (~8-10 min, robust 45s holds) and prints a
`RESULT URL` at the bottom -- send me that one line.

```python
import urllib.request, subprocess, os, glob
URL = "https://gist.githubusercontent.com/dcondrey/8c64aa4bb077cbfb6243df8b7b67d701/raw/selfmeasure.py"
req = urllib.request.Request(URL, headers={"User-Agent": "curl/8"})
open("selfmeasure.py", "wb").write(urllib.request.urlopen(req).read())
env = dict(os.environ, SELFMEASURE_HOLD="45", SELFMEASURE_SETTLE="10")
r = subprocess.run(["python", "selfmeasure.py"], capture_output=True, text=True, env=env)
print(r.stdout[-4000:])
if r.stderr.strip():
    print("STDERR:", r.stderr[-800:])
fs = glob.glob("results/selfmeasure-*.json")
if fs:
    data = open(fs[0]).read()
    try:
        u = urllib.request.urlopen(urllib.request.Request("https://paste.rs/", data=data.encode(), method="POST")).read().decode().strip()
        print("\n\n==== RESULT URL (send me this) ====")
        print(u)
    except Exception as e:
        print("\nupload failed:", e, "\n", data)
```
